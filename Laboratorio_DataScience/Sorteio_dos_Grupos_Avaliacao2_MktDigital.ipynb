```python
import random

# 1. Definição dos horários informados
horarios = [
    "8h20 às 8h40",
    "8h40 às 9h00",
    "9h00 às 9h20",
    "9h20 às 9h40",
    "10h00 às 10h20",
    "10h20 às 10h40"
]

# 2. Definição dos 6 grupos que participarão do sorteio
grupos = ["Grupo 1", "Grupo 2", "Grupo 3", "Grupo 4", "Grupo 5", "Grupo 6"]

print("=== PASSO 1: Horários de apresentação dos grupos: ===")
print("Horários definidos para a avaliação:")
for h in horarios:
    print(f"  - {h}")
print("\nGrupos participantes do sorteio:")
print(", ".join(grupos))

print("\n=== PASSO 2: Sorteio Aleatório ===")
print("O algoritmo embaralha a lista de grupos para que a ordem final seja totalmente imparcial.")
grupos_embaralhados = grupos.copy()
random.shuffle(grupos_embaralhados)
print("Ordem resultante do embaralhamento:")
print(", ".join(grupos_embaralhados))

print("\n=== PASSO 3: Resultado da Associação ===")
print("Por fim, cada grupo embaralhado é mapeado cronologicamente para um dos horários fixados:\n")

cronograma_sorteado = []
for i, horario in enumerate(horarios):
    grupo = grupos_embaralhados[i]
    cronograma_sorteado.append((horario, grupo))
    print(f"[{horario}] -> {grupo}")

print("\nSorteio concluído!")
```
